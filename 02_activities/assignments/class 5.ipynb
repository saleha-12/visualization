```python
#importing necessary libraries
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import PIL
import requests
import scipy
```

```python
#loading our dataset
tips = sns.load_dataset("tips")
print(tips)
```

```python
# creating a line plot of our dataset
sns.lineplot(data = tips,
             x = 'total_bill',
             y = 'tip')
```

```python
# adding gridlines to our plot
sns.set_style('whitegrid')

#sns.set_style('dark')
#sns.set_style('ticks')

sns.lineplot(data = tips,
             x = 'total_bill',
             y = 'tip')
```

```python
# changing our plot titles and labels

tipgraph = sns.lineplot(data = tips,
             x = 'total_bill',
             y = 'tip')

tipgraph.set(title = 'Tips vs Total Bill',
             xlabel = 'Total Bill ($)',
             ylabel = 'Tip Amount ($)')
```

```python
#customizing/modifying our plot
fig = plt.subplots(figsize = (10,3))

tipgraph = sns.lineplot(data = tips,
                        x = 'total_bill',
                        y = 'tip',
                        color = 'hotpink',
                        linestyle = '--',
                        linewidth = 3,
                        marker ='o',
                        markerfacecolor = 'indigo')

tipgraph.set(title = 'Tips vs Total Bill',
             xlabel = 'Total Bill ($)',
             ylabel = 'Tip Amount ($)')
```

```python
# trying scatter plot

tipgraph = sns.scatterplot(data = tips,
                        x = 'total_bill',
                        y = 'tip',
                        style = 'time',
                        hue = 'day',
                        palette = ['purple','hotpink','deepskyblue','yellowgreen'])

tipgraph.set(title = 'Tips vs Total Bill',
             xlabel = 'Total Bill ($)',
             ylabel = 'Tip Amount ($)')
```

```python
#creating pairplot
sns.pairplot(data = tips,
            hue = 'day')
```

```python
# CLASS ACTIVITY(creating and analyzing relplot)
daysplot = sns.relplot(data = tips,
                      x ="total_bill",
                      y="tip",
                      hue="sex",
                      col="day",
                      kind="scatter",
                      col_wrap=2)
```

```python
# working on dynamic data vis
import plotly.graph_objects as go #go is graph objects

x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
```

```python
graph = go.Figure()

graph.add_trace(go.Bar(x=x1, y=y1))
```

```python
# updating our layout to create labels
graph.update_layout(title="Pirate Scores",
                    xaxis_title="Pirates",
                    yaxis_title="Score")

graph.show()
```

```python
#modifying our graph
graph = go.Figure()
graph.add_trace(go.Scatter(x=x1,
                           y=y1,
                           mode='markers',
                           marker=dict(
                               size=15,
                               color='hotpink',
                               opacity=1,
                               line=dict(width=5, color='purple')
                           )))

graph.update_layout(title='Interactive Pirate Plot',
                    xaxis_title='Pirates',
                    yaxis_title='Scores',
                    width=500, height=500)
```

```python
# Working with Wordclouds and venn diagrams
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv", on_bad_lines='skip')
df
```

```python
# join all our text from each row from our quote column into a string
text = " ".join(each for each in df.quote)

# generate our wordcloud image
wordcloud = WordCloud(background_color="white",
                      colormap = 'inferno').generate(text)

# use matplotlib syntax to put our image in a figure
fig, ax = plt.subplots(figsize=(7, 3))
ax.imshow(wordcloud)

ax.axis("off")
         
```

```python
# Venn Diagrams
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted

# defining our sets
A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])
```

```python
# creating a venn diagram
diagram = venn2_unweighted([A, B],
                           set_labels = ('Fruits', 'Pies'),
                           set_colors=("blue", "red"),
                           alpha=0.5)

diagram.get_label_by_id("10").set_text("\n".join(A - B))  #left
diagram.get_label_by_id("11").set_text("\n".join(A & B))  #middle
diagram.get_label_by_id("01").set_text("\n".join(B - A))  #right
```
