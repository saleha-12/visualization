```python
# Importing necessary libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Loading our data
df = pd.read_csv("02_activities/assignments/All sectors and seconded employees 2020.csv")
df
```

```python
# Checking for missing values
missing_values = df.isnull().sum()

# Displaying the number of missing values for each column
missing_values
```

```python
# Converting Salary to numeric (removing dollar signs and commas)
df['Salary'] = df['Salary'].replace('[\$,]', '', regex=True).astype(float)

# Filtering for top 30 most common job titles for clarity
top_job_titles = df['Job title'].value_counts().head(30).index
df_filtered = df[df['Job title'].isin(top_job_titles)]

# Plotting
plt.figure(figsize=(12, 8))
sns.boxplot(x='Job title', y='Salary', data=df_filtered, palette='Set3')
plt.title('Salary Distribution by Job Title (2020)')
plt.xlabel('Job Title')
plt.ylabel('Salary ($)')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
