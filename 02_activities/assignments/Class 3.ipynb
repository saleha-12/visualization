```python
#importing necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import PIL
import requests
import scipy
```

```python
#defining our x and y variables
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

```python
#plotting our defined variables as line plots
fig, ax = plt.subplots(figsize =(5,3))
ax.plot(x,y1)
ax.plot(x,y2)
fig.show
```

```python
#adding a legend to our graph for more clarity
fig, ax = plt.subplots(figsize =(5,3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')
```

```python
#changing the appearance of our legend by adding different arguments
fig, ax = plt.subplots(figsize =(5,3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc = 'lower right',
          frameon = True,
          fontsize = 12,
          ncol = 2,
          shadow = True)
```

```python
# modifying the location of the legend 
fig, ax = plt.subplots(figsize =(5,3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc = 'upper left',
          bbox_to_anchor = (1,1))
```

```python
# adding text and annotations to the plot and plotting a scatter plot now 
fig, ax = plt.subplots(figsize =(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

ax.text(10,95, 'This value is important')
```

```python
# modifying text annotations
fig, ax = plt.subplots(figsize =(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

ax.text(10,95, 'This value is important',
        ha = 'center',
        color = 'red',
        size = 20)
```

```python
# positioning text annotations
fig, ax = plt.subplots(figsize =(5,3))
ax.axis([0,10,0,10])

ax.text(1,5, "Data: (1,5)", transform = ax.transData)
ax.text(0.5,0.1, "Axes: (0.5,0.1)", transform = ax.transAxes)
ax.text(0.2,0.2, "Figure: (0.2,0.2)", transform =fig.transFigure)
```

```python
#annotating eith arrows
fig, ax = plt.subplots(figsize =(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

ax.annotate("This value is important!", xy = (10,95),
        arrowprops = dict(facecolor = 'black', shrink = 0.05))
```

```python
# modifying arrows
fig, ax = plt.subplots(figsize =(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

ax.annotate('This is important!',
            xy=(10,95), xytext=(20,94),
            arrowprops = dict(arrowstyle = "wedge",
                              color = "hotpink"))
```

```python
# removing the tick marks and labels
fig, ax = plt.subplots(figsize =(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter())
```

```python
# limitting the tick marks to only 3 on x-axis
fig, ax = plt.subplots(figsize =(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

ax.yaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
# Tick marks to appear at a particular gap/interval
fig, ax = plt.subplots(figsize =(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

ax.xaxis.set_major_locator(plt.MultipleLocator(5))
```

```python
# Rotating Tick marks
fig, ax = plt.subplots(figsize =(5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

ax.xaxis.set_major_locator(plt.MultipleLocator(5))
plt.xticks(rotation = 45, ha = 'right', fontsize = 12, font = "serif")
```

```python
# CLASS ACTIVITY (modifying axis title fonts)
font1 = {'family': 'serif', 'color': 'indigo'}

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label = "Person 1")

plt.xlabel("Shiny New Axis", fontsize = 18, fontdict =font1)
```

```python
# SUBPLOTS
#importing necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import PIL
import requests
import scipy
```

```python
# generating new data sets
x1 = np.arange(50)
y1 = np.random.randint(0,75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110,180,240,99,220])
```

```python
# calling our two axes and plotting our data into scatter and bar plots
fig, (ax1,ax2) = plt.subplots(ncols = 2,
                              nrows = 1,
                              figsize =(7,3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
```

```python
# CLASS ACTIVITY (customizing our subplots)
fig, (ax1,ax2) = plt.subplots(ncols = 2,
                              nrows = 1,
                              figsize =(7,3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)

ax1.scatter(x1, y1, color='blue')
ax1.set_title("Scatter Plot 1")
ax1.set_xlabel("X1 Axis")
ax1.set_ylabel("Y1 Axis")
ax1.legend()

ax2.bar(x2, y2, color='purple')
ax2.set_title("Bar Plot")
ax2.set_xlabel("X2 Axis")
ax2.set_ylabel("Y2 Axis")
ax2.legend()
```

```python
# creating mosaics
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],
                                    ['ax2', 'ax3']],
                                    figsize = (7,4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
```

```python
# Modifying figure layout to a bit more complex form
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],
                                    ['ax2', 'ax3']],
                                    figsize = (7,4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)


someaxes["ax1"].set_xlabel('A Big Label', fontsize = 18)
someaxes["ax2"].set_xlabel('Another Label', fontsize = 18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize = 18)
```

```python
# fixing the title for scatter plot and line plot
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],
                                    ['ax2', 'ax3']],
                                    figsize = (7,4),
                                    layout = "constrained")
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)

someaxes["ax1"].set_xlabel('A Big Label', fontsize = 18)
someaxes["ax2"].set_xlabel('Another Label', fontsize = 18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize = 18)
```

```python
# introducing new data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110,180,240,99,220])
y2 = np.array([170,100,90,120,50])

fig, ax = plt.subplots(figsize=(7,3))

ax.bar(x,y1,color = 'indigo')
ax.plot(x,y2,color = 'red')
```

```python
#calculating the sd of our new var y2
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color = 'red')

ax.errorbar(x,y2,
            yerr=y2_sd,
            fmt = "none")
```

```python
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color = 'red')

ax.errorbar(x,y2,
            yerr=y2_sd,
            fmt = "none",
            ecolor = "indigo",
            elinewidth = 4,
            capsize = 6,
            capthick = 4)
```

```python
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color = 'red')

ax.errorbar(x,y2,
            yerr=y2_sd,
            fmt = "none",
            ecolor = "indigo",
            elinewidth = 4,
            capsize = 6,
            capthick = 4,
            errorevery = 2)
```
